```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from numpy import dot
from numpy.linalg import norm
```

```python
df = pd.read_csv("data/cleaned/features.csv")
```

```python
y = df["SalePrice"].to_numpy() # target variable
mean_y = np.sum(y) / len(y) # mean of the target variable
std_y = np.sqrt(np.sum((y - mean_y) ** 2) / len(y)) # standard deviation of the target variable
print(f"Mean: {mean_y}") # display result
print(f"Standard Deviation: {std_y}") # display result
# • Compute the mean and standard deviation of the target manually using NumPy (not pandas.mean())
```

```python
X = df["Overall Qual"].to_numpy() # feature variable
mean_X = np.sum(X) / len(X) # mean of the feature variable
std_X = np.sqrt(np.sum((X - mean_X) ** 2) / len(X)) # standard deviation of the feature variable
z_manual = (X - mean_X) / std_X # z-score normalization
print(z_manual[:5]) # display result

scaler = StandardScaler() # create an instance of StandardScaler
z_sklearn = scaler.fit_transform(X.reshape(-1, 1)) # standardize the feature using StandardScaler
z_sklearn = z_sklearn.flatten() # flatten the result to 1D array
print(z_sklearn[:5]) # display result
# • Standardise one column by hand using broadcasting: z = (X - mean) / std — compare with StandardScaler output
```

```python
index_max = df["SalePrice"].idxmax() # index of the row with the maximum SalePrice
index_min = df["SalePrice"].idxmin() # index of the row with the minimum SalePrice

numeric_cols = df.select_dtypes(include="number").columns.drop("SalePrice") # select numeric columns excluding the target variable

vector_max = df.loc[index_max, numeric_cols].to_numpy() # convert the row with the maximum SalePrice to a NumPy array
vector_min = df.loc[index_min, numeric_cols].to_numpy() # convert the row with the minimum SalePrice to a NumPy array

dot_product = dot(vector_max, vector_min) # compute the dot product of the two vectors
norm_max = norm(vector_max) # compute the L2 norm (magnitude) of the vector with the maximum SalePrice
norm_min = norm(vector_min) # compute the L2 norm (magnitude) of the vector with the minimum SalePrice
cos_sim = dot_product / (norm_max * norm_min) # compute the cosine similarity between the two vectors

print(f"Cosine Similarity: {cos_sim}") # display result
# • Compute cosine similarity between the feature vectors of the highest-value and lowest-value records
```

```python
high_quality = df[df["Overall Qual"] >= 7]  # filter rows with high quality (>= 7)
threshold = df["SalePrice"].mean()  # mean SalePrice

count_above = high_quality[high_quality["SalePrice"] > threshold].shape[0]  # count above mean
total_high = high_quality.shape[0]  # total high-quality rows

probability = count_above / total_high if total_high > 0 else 0 # avoid division by zero
print(f"Probability: {probability:.4f}")  # result
# • Estimate a probability: e.g., what fraction of high-quality items have a target above a certain threshold?
```
